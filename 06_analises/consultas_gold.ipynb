```python
# Case iFood — Consultas na camada Gold

# Fonte: `case_ifood.gold.trips_consolidado` (unificação yellow + green, enriquecida com as dimensões TLC). Período: jan–mai 2023. Uma célula SQL por pergunta do enunciado.
```

## Pergunta 1 — Média de valor total (total_amount) recebido em um mês, yellow

> Qual a média de valor total (total_amount) recebido em um mês considerando todos os yellow táxis da frota?

Resposta por mês (jan–mai/2023), taxis yellow, sobre a gold:

```python
%sql
SELECT date_format(pickup_datetime, 'yyyy-MM') AS mes,
       ROUND(AVG(total_amount), 2) AS media_total_amount,
       COUNT(*) AS corridas
FROM case_ifood.gold.trips_consolidado
WHERE taxi_type = 'yellow'
GROUP BY 1
ORDER BY 1
```

## Pergunta 2 — Média de passageiros por hora do dia, maio, frota toda

> Qual a média de passageiros (passenger_count) por cada hora do dia que pegaram táxi no mês de maio considerando todos os táxis da frota?

Maio/2023, yellow + green (frota toda), por hora da pickup:

```python
%sql
SELECT hour(pickup_datetime) AS hora_do_dia,
       ROUND(AVG(passenger_count), 2) AS media_passageiros,
       COUNT(*) AS corridas
FROM case_ifood.gold.trips_consolidado
WHERE month(pickup_datetime) = 5
GROUP BY 1
ORDER BY 1
```
